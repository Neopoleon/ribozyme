```python
import urllib
import glob
import requests
from bs4 import BeautifulSoup
from tqdm import tqdm
```

```python
def get_bprnaID(file):
    return 'bpRNA_' + file.split('bpRNA_')[-1].split('.bpseq')[0]

def get_link(file):
    return 'https://bprna.cgrb.oregonstate.edu/search.php?query=' + get_bprnaID(file)

def get_links(dir):
    files = glob.glob(dir + '/*.bpseq')
    return [get_link(file) for file in files]

def get_RNA_type(file):
    url = get_link(file)
    response = requests.get(url)
    rna_type = response.text.split('RNA Type:</b>')[1].split('<br>')[0].strip()
    return rna_type

def get_RNA_types(dir):
    files = glob.glob(dir + '/*.bpseq')
    types = []
    for file in tqdm(files):
        try:
            types.append(get_RNA_type(file))
        except:
            types.append(None)
    return types
```
