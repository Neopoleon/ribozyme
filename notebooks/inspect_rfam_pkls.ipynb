# Inspect RFAM Pickle Files

This notebook helps you explore the RFAM metadata labels stored in pickle files.

```python
import pickle
import pandas as pd
from pathlib import Path
```

## Load RFAM Pickle Files

```python
# Load rfam_types.pkl
with open('../rfam/rfam_types.pkl', 'rb') as f:
    rfam_types = pickle.load(f)

print("rfam_types.pkl loaded")
print(f"Type: {type(rfam_types)}")
print(f"Length: {len(rfam_types) if hasattr(rfam_types, '__len__') else 'N/A'}")
```

```python
# Inspect structure
if isinstance(rfam_types, dict):
    print("\nFirst 10 entries:")
    for i, (k, v) in enumerate(list(rfam_types.items())[:10]):
        print(f"{k}: {v}")
elif isinstance(rfam_types, list):
    print("\nFirst 10 entries:")
    print(rfam_types[:10])
else:
    print("\nContent:")
    print(rfam_types)
```

```python
# Load rfam_types_full.pkl
with open('../rfam/rfam_types_full.pkl', 'rb') as f:
    rfam_types_full = pickle.load(f)

print("rfam_types_full.pkl loaded")
print(f"Type: {type(rfam_types_full)}")
print(f"Length: {len(rfam_types_full) if hasattr(rfam_types_full, '__len__') else 'N/A'}")
```

```python
# Inspect structure
if isinstance(rfam_types_full, dict):
    print("\nFirst 10 entries:")
    for i, (k, v) in enumerate(list(rfam_types_full.items())[:10]):
        print(f"{k}: {v}")
elif isinstance(rfam_types_full, list):
    print("\nFirst 10 entries:")
    print(rfam_types_full[:10])
else:
    print("\nContent:")
    print(rfam_types_full)
```

```python
# Load rfam_errors.pkl
with open('../rfam/rfam_errors.pkl', 'rb') as f:
    rfam_errors = pickle.load(f)

print("rfam_errors.pkl loaded")
print(f"Type: {type(rfam_errors)}")
print(f"Length: {len(rfam_errors) if hasattr(rfam_errors, '__len__') else 'N/A'}")
```

```python
# Inspect errors
if isinstance(rfam_errors, dict):
    print("\nSample entries:")
    for i, (k, v) in enumerate(list(rfam_errors.items())[:10]):
        print(f"{k}: {v}")
elif isinstance(rfam_errors, list):
    print("\nSample entries:")
    print(rfam_errors[:10])
else:
    print("\nContent:")
    print(rfam_errors)
```

## Summary Statistics

```python
# Create summary if data is dictionary format
if isinstance(rfam_types, dict):
    print(f"Total entries in rfam_types: {len(rfam_types)}")
    
    # Get unique values if values are strings
    if rfam_types and isinstance(list(rfam_types.values())[0], str):
        unique_types = set(rfam_types.values())
        print(f"\nUnique RNA types: {len(unique_types)}")
        print("\nRNA type distribution:")
        type_counts = pd.Series(list(rfam_types.values())).value_counts()
        print(type_counts)
```

## Example: Look up a specific sequence

The .st files contain 4 lines per sequence:
1. Name/ID
2. Sequence
3. Dot-bracket structure
4. Structural annotation

```python
# Example: Look up a sequence by name
if isinstance(rfam_types, dict):
    example_key = list(rfam_types.keys())[0] if rfam_types else None
    if example_key:
        print(f"Example lookup: {example_key}")
        print(f"RNA type: {rfam_types[example_key]}")
```
